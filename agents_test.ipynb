```python
import inspect
from typing import Callable, Any, List, Dict, get_type_hints
from openai import OpenAI
import json


class Tool:
    """Invólucro que transforma funções Python em ferramentas com JSON Schema puro."""
    def __init__(self, func: Callable):
        self.func = func
        self.name = func.__name__
        self.description = (inspect.getdoc(func) or "").strip()
        self.__doc__ = func.__doc__

        # Mapeamento simples de tipos Python para JSON Schema
        type_mapping = {
            int: "integer",
            float: "number",
            str: "string",
            bool: "boolean",
            list: "array",
            dict: "object",
            List: "array",
            Dict: "object"
        }

        sig = inspect.signature(func)
        type_hints = get_type_hints(func)
        
        properties = {}
        required = []

        # Constrói o schema iterando sobre os parâmetros
        for param_name, param in sig.parameters.items():
            # Ignora *args e **kwargs
            if param.kind in (inspect.Parameter.VAR_POSITIONAL, inspect.Parameter.VAR_KEYWORD):
                continue

            # Pega o tipo (default para string caso não tenha anotação)
            param_type = type_hints.get(param_name, str)
            json_type = type_mapping.get(param_type, "string")

            properties[param_name] = {
                "type": json_type
            }

            # Se não tem valor padrão, o parâmetro é obrigatório
            if param.default == inspect.Parameter.empty:
                required.append(param_name)

        # Monta a estrutura exata que a API espera
        self.schema = {
            "type": "function",
            "function": {
                "name": self.name,
                "description": self.description,
                "parameters": {
                    "type": "object",
                    "properties": properties,
                    "required": required
                }
            }
        }

    def __call__(self, *args, **kwargs):
        return self.func(*args, **kwargs)


def tool(func: Callable) -> Tool:
    return Tool(func)


# class QwenChatModel:
#     """Wrapper leve para o client do OpenAI com suporte a bind_tools."""
#     def __init__(self, model_name: str, client: OpenAI, temperature: float = 0, tools: List[Tool] = None):
#         self.model_name = model_name
#         self.client = client
#         self.temperature = temperature
#         self.tools = tools or []

#     def bind_tools(self, tools: List[Tool]) -> "QwenChatModel":
#         """Retorna uma nova instância do modelo vinculada às ferramentas."""
#         return QwenChatModel(
#             model_name=self.model_name,
#             client=self.client,
#             temperature=self.temperature,
#             tools=tools
#         )

#     def invoke(self, messages: List[Dict[str, Any]]) -> Any:
#         payload = {
#             "model": self.model_name,
#             "messages": messages,
#             "temperature": self.temperature,
#         }
#         if self.tools:
#             payload["tools"] = [t.schema for t in self.tools]

#         return self.client.chat.completions.create(**payload)
```

```python
from abc import ABC, abstractmethod
from openai import OpenAI
from typing import List, Dict, Any

import json
import re
import uuid



import os
from typing import Dict, Any

@tool
def verificar_arquivos() -> Dict[str, Any]:
    """
    Lista todos os arquivos disponíveis no diretório atual.
    """
    try:
        # Obtém todos os itens do diretório atual e filtra apenas o que for arquivo
        todos_itens = os.listdir(".")
        arquivos = [f for f in todos_itens if os.path.isfile(f)]
        
        return {
            "status": "sucesso",
            "total_arquivos": len(arquivos),
            "arquivos": arquivos
        }
    except Exception as e:
        return {
            "status": "erro",
            "mensagem": f"Erro ao acessar a pasta: {str(e)}"
        }

@tool
def read_start_file(file_name: str, max_caracteres: int = 2000) -> Dict[str, Any]:
    """
    Lê o trecho inicial de um arquivo para não estourar a memória/contexto do LLM.

    Args :  file_name: str = Nome do arquivo presente nessa pasta que se quer ler
            max_caracteres: int = Número de caracteres máximos lidos a partir do inicio ( está setado com valor padrão de 2000 para caso não seja atribuído um valor)
    """
    try:
        # Checa se o arquivo realmente existe no disco
        if not os.path.exists(file_name):
            return {
                "status": "erro", 
                "mensagem": f"O arquivo '{file_name}' não foi encontrado no diretório atual."
            }
        
        if os.path.isdir(file_name):
            return {
                "status": "erro", 
                "mensagem": f"'{file_name}' é uma pasta, não um arquivo."
            }

        # Lê os primeiros 'max_caracteres' do arquivo com encoding seguro
        with open(file_name, "r", encoding="utf-8", errors="replace") as f:
            conteudo = f.read(max_caracteres)

        return {
            "status": "sucesso",
            "file_name": file_name,
            "conteudo_inicial": conteudo,
            "troncado": len(conteudo) >= max_caracteres
        }
        
    except Exception as e:
        return {
            "status": "erro",
            "mensagem": f"Erro ao ler o arquivo '{file_name}': {str(e)}"
        }

# Dicionário útil para disparar a execução no seu loop ReAct
AVAILABLE_TOOLS = {
    "verificar_arquivos": verificar_arquivos,
    "read_start_file": read_start_file
}

class Ai_Agent_simple_loop():
    """Apenas para testar novas LLMs e entender como é o comportamento destas"""
    def __init__(self , model = "qwen3.6:35b-a3b" ):
        self.model = model

    def set_llm_gateway(self, base_url_="http://localhost:11434/v1", api_key_="bifrost"):
        """Define ou atualiza o cliente explicitamente."""
        # Salvamos direto no dicionário interno para evitar recursão infinita
        print("[Gateway] Instanciado via método Setter")
        
        self.__dict__['client'] = OpenAI(
            base_url=base_url_,
            api_key=api_key_
        )

    def __getattr__(self, name):
        """Intercepta a busca por atributos que NÃO existem no objeto."""
        # Qualquer adição de comportamento pode ser obtida se colocando este novo comportamento em uma classe filha e realizando a chamada de função super()
        if name == 'client':
            # Se alguém tentou usar o 'client' sem antes chamar o set_llm_gateway,
            # nós criamos ele aqui com os valores padrão de fábrica.
            print("[Gateway] Inicializando cliente padrão via Fallback...")
            
            # Instancia e injeta diretamente no objeto para as próximas chamadas
            self.__dict__['client'] = OpenAI(
                base_url="http://localhost:11434/v1",
                api_key="bifrost"
            )
            return self.__dict__['client']
        elif name == "system_prompt" :
            self.__dict__[ "system_prompt" ] = self.format_system_prompt()

            return self.__dict__[ "system_prompt" ]
        elif name == "tools" :
            # Em teoria no futuro posso colocar o mecanismo de adicionar novas tools ( inclusive criadas pelo próprio agent )
            self.__dict__[ "tools" ] = {}
            self.tool_registry()
            return self.__dict__[ "tools" ]
            
        elif name == "tools_definitions" :
            # Em teoria no futuro posso colocar o mecanismo de adicionar novas tools ( inclusive criadas pelo próprio agent )
            self.__dict__[ "tools_definitions" ] = []
            self.tool_registry()
            return self.__dict__[ "tools_definitions" ]


        
        # Mantém o comportamento padrão do Python para outros atributos inexistentes
        raise AttributeError(f"'{type(self).__name__}' object has no attribute '{name}'")

    def start_chat(self , max_number_of_steps_for_a_response = 5 ):
        loop_again = True

        print("Vamos iniciar o loop do Agent ")

        chat_hist = [ {"role" : "system",
            "content": self.system_prompt
        }
        ]

        while loop_again :

            # Bloco de Interação textual do Usuário :
            tools_called = [None]
            total_steps_available = max_number_of_steps_for_a_response
            user_prompt  = input("Digite seu prompt aqui : ")


            print(f"\nUser : {user_prompt}")
            # loop_again = False
            if user_prompt.upper() == "close".upper():
                loop_again = False
                print("\n___::::CONVERSA ENCERRADA::::___")
                break
            
            chat_hist = self.generate_message(  chat_hist , user_prompt )

            # Bloco de Processamento do Agent :

            # 2. Feedback visual forçado para você saber que a inferência começou
            print("\nAI está pensando...\n", end="", flush=True)

            response_raw = None
            analized_response = None
            steps_avaiable = True
            while steps_avaiable :
                # Depois separar melhor essa parte para exibir Tudo como se fosse uma resposta Só em várias Etapas
                response_raw = self.generate_response( chat_hist , tools_ = self.tools_definitions ) # , tools_ = my_tools 
                chat_hist = self.generate_message(  chat_hist , response_raw.content , "assistant" )

                # function_call=None, tool_calls=None
                print(f"\nResponse_Raw.function_call dentro do Loop : {response_raw.function_call }\n")
                print(f"\nResponse_Raw.tool_calls    dentro do Loop : {response_raw.tool_calls }\n")

                break
                
                

            #__Se Esses Número estiverem diferentes é porquê ouve chamada de tool
            # if total_steps_available != max_number_of_steps_for_a_response and :
            #     response = analized_response
            # else :
            #     response = response_raw


            if total_steps_available < 1 :
                response = response_raw #analized_response
            else :
                response = response_raw



            print(f"\nAI : Content : {response.content}\n\n     tool_calls : {response.tool_calls}" , flush=True)

            chat_hist = self.generate_message(  chat_hist , response , "assistant" )

    def format_system_prompt(self, prompt_base: str = None ) -> str:
        """Formata o system prompt específico para o modelo e o Retorna no final ."""
        
        self.system_prompt = """Você é um Agente Autônomo Especialista (nível Engenheiro Principal de Software), projetado para resolver problemas complexos de forma metódica, autônoma e implacável.

            Sua arquitetura mental é baseada no ciclo "Pensar -> Agir -> Avaliar". Para cada solicitação do usuário, você deve planejar estrategicamente antes de tomar decisões.

            ====================================================================
            DIRETRIZES DE EXECUÇÃO E RACIOCÍNIO (CHAIN-OF-THOUGHT)
            ====================================================================
            1. PLANEJAMENTO PRÉVIO: Antes de emitir a resposta final, avalie a complexidade do problema. Se o problema for complexo, divida-o em etapas lógicas menores.
            2. VERIFICAÇÃO DE FATOS (EVIDÊNCIAS): NUNCA presuma o conteúdo de um arquivo, a estrutura de um diretório ou o estado de um sistema. Se você não tem certeza absoluta, você DEVE usar as ferramentas disponíveis para investigar e confirmar os dados antes de prosseguir.
            3. ADAPTAÇÃO A ERROS: Se uma chamada de ferramenta retornar um erro ou um resultado inesperado, não entre em pânico e não desista. Analise a mensagem de erro logicamente, ajuste os parâmetros e faça uma nova tentativa.
            4. MINIMIZAÇÃO DE ALUCINAÇÕES: Se as ferramentas disponíveis não forem suficientes para resolver o problema, ou se você não tiver acesso à informação necessária, seja totalmente transparente com o usuário e explique exatamente o que está bloqueando o seu avanço.
            5. PROIBIÇÃO DE EXEMPLOS JSON: NUNCA escreva blocos de código JSON simulando chamadas de ferramentas no meio da sua resposta para o usuário. Se for explicar como uma ferramenta funciona, explique apenas em texto natural, sem usar a sintaxe {"name":...}.
            6. PROIBIÇÃO DE INVENÇÃO: Você só pode usar as ferramentas que foram explicitamente fornecidas a você. NUNCA invente nomes de ferramentas (como 'responder_usuario', 'final_answer', etc).
            7. COMO RESPONDER AO USUÁRIO: Quando você decidir que tem a resposta final, NÃO chame nenhuma ferramenta. Simplesmente escreva a sua resposta em texto puro (Markdown). A ausência de uma chamada de ferramenta é o gatilho que envia a sua mensagem ao usuário.
            
            ====================================================================
            DIRETRIZES DE USO DE FERRAMENTAS (TOOL CALLING)
            ====================================================================
            - Você possui acesso a um conjunto de ferramentas (Tools) perfeitamente integradas ao seu ambiente.
            - Utilize-as de forma proativa. Você pode (e deve) encadear múltiplas ferramentas em sequência se a tarefa exigir investigação profunda.
            - Escolha os argumentos das ferramentas com precisão cirúrgica, baseando-se estritamente no histórico da conversa e nos dados que você já coletou.

            ====================================================================
            DIRETRIZES DE COMUNICAÇÃO
            ====================================================================
            - Idioma: Responda sempre em Português do Brasil (pt-BR) de forma técnica, culta e direta.
            - Tom: Profissional, pragmático e focado em soluções. Evite saudações excessivas ou linguagem poética/emocional.
            - Formatação: Quando estiver redigindo a resposta final para o usuário, use formatação Markdown rica (tabelas, listas, blocos de código com linguagem definida) para tornar a informação o mais legível e estruturada possível.
            - Síntese: Ao entregar a solução final, referencie os dados que você descobriu usando suas ferramentas para embasar suas conclusões.

            Seu objetivo final é resolver a dor do usuário com o menor número de interações manuais possíveis, automatizando a investigação e a resolução através da sua inteligência e das suas ferramentas.
            """
        
        return self.system_prompt

    def tool_registry(self , tools_to_add : dict = {} , tools_to_add_definitions : list[dict] = [] ) -> list[dict]:
        """Dicionário obrigatório para mapear as ferramentas que podem ser chamadas pelo Agent."""
        
        # 0. Garante a existência segura das variáveis internas sem acionar o __getattr__
        if "tools_definitions" not in self.__dict__:
            self.__dict__["tools_definitions"] = []
        if "tools" not in self.__dict__:
            self.__dict__["tools"] = {}

        # === 1. REGISTRO DOS JSON SCHEMAS (DEFINIÇÕES) ===
        registred_tools = self.__dict__["tools_definitions"]
        my_tools = [
            {
                "type": "function",
                "function": {
                    "name": "verificar_arquivos",
                    "description": "Nomeia todos os arquivos disponíveis na pasta atual e retorna como uma lista",
                    "parameters": {
                        "type": "object",
                        "properties": {},
                        "required": []
                    }         
                }
            }, 
            {
                "type": "function",
                "function": {
                    "name": "read_start_file",
                    "description": "Dado um arquivo selecionado ele lê o conteúdo inicial do arquivo",
                    "parameters": {
                        "type": "object",
                        "properties": {
                            "file_name": {
                                "type": "string",
                                "description": "Nome do arquivo a ser lido"
                            } 
                        },
                        "required": [ "file_name" ]
                    }
                }
            } 
        ]

        # Junta todas as listas de definições e remove duplicatas
        all_tools_combined = registred_tools + my_tools + tools_to_add_definitions
        unique_tools = {}
        for tool in all_tools_combined:
            nome_da_tool = tool["function"]["name"]
            unique_tools[nome_da_tool] = tool
            
        self.__dict__["tools_definitions"] = list(unique_tools.values())


        # === 2. REGISTRO DOS PONTEIROS DE EXECUÇÃO DAS FUNÇÕES ===
        # Registra as funções nativas (Garante que não vão ser apagadas)
        self.__dict__["tools"]["verificar_arquivos"] = verificar_arquivos
        self.__dict__["tools"]["read_start_file"] = read_start_file

        # Adiciona APENAS as novas ferramentas vindas de fora sem sobrescrever as nativas
        for nome_tool, func in tools_to_add.items():
            if nome_tool not in self.__dict__["tools"]:
                self.__dict__["tools"][nome_tool] = func

        return self.tools

    def generate_message(self , chat_hist_ , new_msg , role = "user"):
        chat_hist_ += [{"role":f"{role}" , "content" :f"{new_msg}"}]
        return chat_hist_

    def generate_response( self , messages: list[Dict] , tools_ = [] ) -> str:
        """Call LLM to get response"""

        if len( tools_ ) == 0 :
            response  =  self.client.chat.completions.create(
                model=  self.model , #"qwen2.5-coder:32b",
                # model= "deepseek-r1:32b" ,
                messages = messages,
                temperature = 0.1
            )
        else :
            response  =  self.client.chat.completions.create(
                model = self.model , #"qwen2.5-coder:32b",
                # model= "deepseek-r1:32b" ,
                messages = messages,
                tools = tools_ ,
                temperature = 0.1
            )

        return response.choices[0].message#.content
    

class AIAgentInterface(ABC):

    def set_llm_gateway(self, base_url_="http://localhost:11434/v1", api_key_="bifrost"):
        """Define ou atualiza o cliente explicitamente."""
        # Salvamos direto no dicionário interno para evitar recursão infinita
        print("[Gateway] Instanciado via método Setter")
        
        self.__dict__['client'] = OpenAI(
            base_url=base_url_,
            api_key=api_key_
        )

    def __getattr__(self, name):
        """Intercepta a busca por atributos que NÃO existem no objeto."""
        # Qualquer adição de comportamento pode ser obtida se colocando este novo comportamento em uma classe filha e realizando a chamada de função super()
        if name == 'client':
            # Se alguém tentou usar o 'client' sem antes chamar o set_llm_gateway,
            # nós criamos ele aqui com os valores padrão de fábrica.
            print("[Gateway] Inicializando cliente padrão via Fallback...")
            
            # Instancia e injeta diretamente no objeto para as próximas chamadas
            self.__dict__['client'] = OpenAI(
                base_url="http://localhost:11434/v1",
                api_key="bifrost"
            )
            return self.__dict__['client']
        elif name == "system_prompt" :
            self.__dict__[ "system_prompt" ] = self.format_system_prompt()

            return self.__dict__[ "system_prompt" ]
        elif name == "tools" :
            # Em teoria no futuro posso colocar o mecanismo de adicionar novas tools ( inclusive criadas pelo próprio agent )
            self.__dict__[ "tools" ] = {}
            self.tool_registry()
            return self.__dict__[ "tools" ]
            
        elif name == "tools_definitions" :
            # Em teoria no futuro posso colocar o mecanismo de adicionar novas tools ( inclusive criadas pelo próprio agent )
            self.__dict__[ "tools_definitions" ] = []
            self.tool_registry()
            return self.__dict__[ "tools_definitions" ]


        
        # Mantém o comportamento padrão do Python para outros atributos inexistentes
        raise AttributeError(f"'{type(self).__name__}' object has no attribute '{name}'")

    def start_chat(self , max_number_of_steps_for_a_response = 5 ):
        loop_again = True

        print("Vamos iniciar o loop do Agent ")

        chat_hist = [ {"role" : "system",
            "content": self.system_prompt
        }
        ]

        while loop_again :

            # Bloco de Interação textual do Usuário :
            tools_called = [None]
            total_steps_available = max_number_of_steps_for_a_response
            user_prompt  = input("Digite seu prompt aqui : ")


            print(f"\nUser : {user_prompt}")
            # loop_again = False
            if user_prompt.upper() == "close".upper():
                loop_again = False
                print("\n___::::CONVERSA ENCERRADA::::___")
                break
            
            chat_hist = self.generate_message(  chat_hist , user_prompt )

            # Bloco de Processamento do Agent :

            # 2. Feedback visual forçado para você saber que a inferência começou
            print("\nAI está pensando...\n", end="", flush=True)

            response_raw = None
            analized_response = None
            steps_avaiable = True
            while steps_avaiable :
                # Depois separar melhor essa parte para exibir Tudo como se fosse uma resposta Só em várias Etapas
                response_raw = self.generate_response( chat_hist , tools_ = self.tools_definitions ) 
                chat_hist = self.generate_message(  chat_hist , response_raw.content , "assistant" )

                print(f"\nResponse_Raw dentro do Loop : {response_raw}\n")

                # Se parser vai ser via .content ou .tool_calls depende do suporte do modelo .
                tools_called  = self.parse_response( response_raw )
                if len( tools_called ) > 0 : 

                    print(f"\nProcesso quebrado em partes : Etapa { max_number_of_steps_for_a_response - total_steps_available +1 }\n")

                    tools_calls_results  = self.execute_tool_chain( tools_called )
                    tool_calls_responses = self.build_response_of_tools_results( tools_calls_results )
                    # O último total_steps_available a ser processado é no menor caso igual a 1 
                    analized_response = self.analize_calls_responses( chat_hist , tool_calls_responses , total_steps_available )
                    
                    #Talvez seja provisório :
                    chat_hist = self.generate_message(  chat_hist , analized_response , "user" )
                    # chat_hist = self.generate_message(  chat_hist , analized_response.content , "user" )

                    total_steps_available += -1
                else :
                    break
                
                if total_steps_available < 0 or len( tools_called ) == 0 :
                # if total_steps_available < 1 or len( tools_called ) == 0 :
                    steps_avaiable = False

            #__Se Esses Número estiverem diferentes é porquê ouve chamada de tool
            # if total_steps_available != max_number_of_steps_for_a_response and :
            #     response = analized_response
            # else :
            #     response = response_raw


            if total_steps_available < 1 :
                response = response_raw #analized_response
            else :
                response = response_raw



            print(f"\nAI : Content : {response.content}\n\n     tool_calls : {response.tool_calls}" , flush=True)

            chat_hist = self.generate_message(  chat_hist , response , "assistant" )
            
    def analize_calls_responses( self, chat_hist_ : list[dict] , tool_calls_responses : str  , total_steps_available : int ) -> str:
        """Faz a LLM analisar as respostas das chamadas de função deacordo com o número interações disponíveis no loop de análise """
        
        context_response = ""


        if total_steps_available > 0 :
            


            context_response = f"""\n\n
            ============================================================================================================
            [AMBIENTE DE RACIOCÍNIO INTERNO , AVALIAÇÃO DE RESULTADOS NO MODO DE ANÁLISE INVISÍVEL E TOMADA DE DECISAO ]
            ============================================================================================================


            ORÇAMENTO DE ITERAÇÕES:
            - Ciclos (steps) restantes neste loop: {total_steps_available}

            RESULTADOS OBTIDOS NESTE STEP:
            ---
            {tool_calls_responses}
            ---

            DIRETRIZ DE DECISÃO E EXECUÇÃO (OU EXCLUSIVO - XOR):
            Avalie os resultados acima em relação ao pedido original do usuário. Você DEVE escolher apenas UM dos caminhos abaixo:

            ► CAMINHO A: INVESTIGAÇÃO CONTÍNUA (Necessito de mais dados)
            [Condição]: Escolha esta opção se a resposta final ainda não pode ser dada e você precisa chamar mais ferramentas.
            [Ação]: Ao Usar este espaço você estará em um espaço de rascunho interno (Scratchpad). O texto que você gerar agora NÃO será visto pelo usuário final. Ele servirá exclusivamente como memória de trabalho para a sua própria inteligência na construção da resposta definitiva.

            DIRETRIZES DE EXECUÇÃO (Cumpra todos os passos abaixo metodicamente):
            
            1. ANÁLISE TÉCNICA:
            Analise os resultados das ferramentas fornecidos acima. Extraia os fatos, consolide as informações úteis e cruze os dados de forma técnica. Guarde as fontes de cada informação, pois você precisará referenciá-las no futuro.

            2. DEPURAÇÃO DE FALHAS (DEBUGGING):
            Se o resultado indicar que alguma chamada de ferramenta falhou (erro de sintaxe, parser ou execução), acesse seu histórico, identifique exatamente qual foi o seu erro de formatação na tentativa anterior e escreva explicitamente como você deve corrigir os parâmetros para a próxima tentativa , após descrever em detalhes o como deve ser corrigido use a sua descrição como base para escrever a chamada de forma correta logo a diante 
            
            3. PLANEJAMENTO DE PRÓXIMO PASSO:
            Com base na sua análise e no número de ciclos restantes, determine o estado da sua estratégia. Se você precisar de mais dados, detalhe o planejamento do(s) usos das próxima(s) chamada(s) de ferramentas que fará na sua próxima iteração. Neste detalhamento deve haver a descrição de como serão usadas as tool_calls e após a descrição devem estar na ordem respectiva de sua primeira mensão a chamada em si da tool respeitando metodicamente a sintaxe de uso da tool referenciada.

            4. EXECUÇÃO OBRIGATÓRIA: Após a análise, você DEVE emitir o JSON da próxima ferramenta.
            
            [RESTRIÇÕES ABSOLUTAS da Rota A ]:
            - NÃO escreva saudações conversacionais (ex: "Olá, aqui estão os dados...").
            - NÃO formule a resposta final ao usuário nesta mensagem.
            - Concentre-se exclusivamente em pensar como um investigador técnico organizando os dados e preparando a próxima ação.



            ► CAMINHO B: RESOLUÇÃO E RESPOSTA FINAL AO USUÁRIO
            [Condição]: Escolha esta opção se os resultados lidos já são 100% suficientes para resolver a solicitação do usuário ou você julgar que precisa que o usuário passe alguma nova informação em forma de resposta a uma pergunta sua para continuar a investigação.
            [Ação]: ABANDONE O MODO SCRATCHPAD. O texto que você gerar com a marcação de resposta final agora será exibido DIRETAMENTE para o usuário final. Para garantir que seu raciocínio lógico (CoT) não vaze para o usuário final, você DEVE estruturar sua saída EXATAMENTE com as seguintes tags XML:

            <scratchpad>
            Escreva aqui suas considerações finais, cruzamento de dados, análises técnicas ou conclusão lógica (Invisível para o usuário).
            </scratchpad>
            <final_answer>
            Assuma sua persona de Agente Especialista. Redija a resposta definitiva, clara e amigável (ou a pergunta técnica necessária de ser respondida pelo usuário), formatada em Markdown, referenciando os dados para embasar suas afirmações. O usuário lerá APENAS o que estiver aqui dentro.
            </final_answer>

            [Restrição B]: VOCÊ ESTÁ PROIBIDO DE EMITIR QUALQUER CHAMADA DE FERRAMENTA (JSON) NESTA ROTA. A ausência de JSON é o gatilho que avisa o sistema que o seu texto é a resposta final.

            """


        
        else :


            context_response = f"""\n\n
            ====================================================================
            [ALERTA DO SISTEMA ORQUESTRADOR: LIMITE DE CICLOS INTERNOS ATINGIDO]
            ====================================================================

            Atenção: Você atingiu o número máximo de iterações automáticas permitidas para este ciclo de raciocínio. O modo de investigação interna acabou. 
            A resposta que você gerar a partir deste exato momento SERÁ EXIBIDA DIRETAMENTE PARA O USUÁRIO FINAL.

            ESTRITAMENTE PROIBIDO: Você está terminantemente proibido de solicitar o uso de qualquer nova ferramenta (tool call) nesta iteração. Não emita nenhum JSON de ferramenta.

            Abaixo estão os resultados extraídos das últimas chamadas de ferramentas que você solicitou:
            ---
            {tool_calls_responses}
            ---

            DIRETRIZ DE DECISÃO (OU EXCLUSIVO - XOR):
            Ao analisar os resultados acima e o histórico da conversa, você DEVE adotar APENAS UMA das duas posturas a seguir. É fundamental que você escolha apenas um caminho:

            ► OPÇÃO A: CONCLUSÃO E RESPOSTA FINAL
            [Condição]: Escolha esta opção SE você já coletou informações suficientes para responder à solicitação original do usuário de forma satisfatória, ou se faltam apenas detalhes mínimos que não impedem a resolução do problema central.
            [Ação]: Redija a resposta definitiva. Sintetize todo o conhecimento construído, referencie os resultados coletados nas ferramentas para embasar suas afirmações e entregue a solução clara e coesa ao usuário.

            ► OPÇÃO B: RELATÓRIO DE PROGRESSO E PLANO DE AÇÃO (FEEDBACK INTERMEDIÁRIO)
            [Condição]: Escolha esta opção SE a tarefa for muito complexa e as informações atuais forem fundamentalmente insuficientes para dar uma resposta correta, exigindo obrigatoriamente mais ciclos de coleta de dados.
            [Ação]: Não tente inventar uma resposta final. Em vez disso, comunique-se com o usuário fornecendo um feedback de status transparente. Sua resposta deve conter obrigatoriamente:
            1. O que já foi alcançado: Explique quais etapas da sua estratégia já foram concluídas com sucesso neste loop.
            2. O obstáculo atual: Mostre por que ainda não alcançamos o resultado final (ex: uma tool falhou e precisa ser reescrita, ou uma paginação profunda é necessária).
            3. O Plano de Ação: Liste detalhadamente quais são os próximos passos lógicos. 
            4. Estimativa: Explique ao usuário quantos passos/loops adicionais você estima que serão necessários para resolver o problema em definitivo, pedindo implicitamente a permissão dele para continuar o processo na próxima iteração.

            Lembre-se: Você está falando com o usuário humano agora. Escolha o caminho A ou o caminho B, seja claro, assertivo e não inclua jargões de depuração do sistema.
            
            FORMATAÇÃO OBRIGATÓRIA DA SAÍDA:
            Independentemente de escolher a OPÇÃO A ou OPÇÃO B, para evitar que seus jargões de depuração vazem, você DEVE separar seu raciocínio da resposta usando as seguintes tags XML:

            <scratchpad>
            Sua avaliação final dos dados, reflexão sobre o que faltou ou planejamento interno (Invisível para o usuário).
            </scratchpad>
            <final_answer>
            A mensagem final (seja a resposta da Opção A ou o relatório da Opção B) direcionada diretamente ao usuário.
            </final_answer>

            """


        return context_response 



    def generate_message(self , chat_hist_ , new_msg , role = "user"):
        chat_hist_ += [{"role":f"{role}" , "content" :f"{new_msg}"}]
        return chat_hist_

    def generate_response( self , messages: list[Dict] , tools_ = [] ) -> str:
        """Call LLM to get response"""

        if len( tools_ ) == 0 :
            response  =  self.client.chat.completions.create(
                model=  self.model , #"qwen2.5-coder:32b",
                # model= "deepseek-r1:32b" ,
                messages = messages,
                temperature = 0.1
            )
        else :
            response  =  self.client.chat.completions.create(
                model = self.model , #"qwen2.5-coder:32b",
                # model= "deepseek-r1:32b" ,
                messages = messages,
                tools = tools_ ,
                temperature = 0.1
            )

        return response.choices[0].message#.content
    

    @property
    @abstractmethod
    def tool_registry(self , tools_to_add : list[dict] = [] ) -> list[dict] :
        """Dicionário obrigatório para mapear as ferramentas que podem ser chamadas pelo Agent."""
        pass

    @abstractmethod
    def format_system_prompt(self, prompt_base: str = None ) -> str:
        """Formata o system prompt específico para o modelo e o Retorna no final ."""
        pass

    @abstractmethod
    def execute_tool_chain(self, tools: list ) -> dict:
        """Define como as ferramentas serão encadeadas e chamadas."""
        pass

    @abstractmethod
    def parse_response(self, raw_response: str) -> list[dict] :
        """Processa a saída bruta do modelo (JSON, texto, etc) e retorna uma lista de todas as tools_calls."""
        pass
    @abstractmethod
    def build_response_of_tools_results( self, tool_responses : dict ) -> str :
        """Monta a resposta textual que consolida os resultados das chamadas de ferramentas"""
        pass
    
    


class qwen2_5_coder_Agent( AIAgentInterface ):

    def __init__(self , model = "qwen2.5-coder:32b" ):
        super().__init__()
        self.model = model

    def tool_registry(self , tools_to_add : dict = {} , tools_to_add_definitions : list[dict] = [] ) -> list[dict]:
        """Dicionário obrigatório para mapear as ferramentas que podem ser chamadas pelo Agent."""
        
        # 0. Garante a existência segura das variáveis internas sem acionar o __getattr__
        if "tools_definitions" not in self.__dict__:
            self.__dict__["tools_definitions"] = []
        if "tools" not in self.__dict__:
            self.__dict__["tools"] = {}

        # === 1. REGISTRO DOS JSON SCHEMAS (DEFINIÇÕES) ===
        registred_tools = self.__dict__["tools_definitions"]
        my_tools = [
            {
                "type": "function",
                "function": {
                    "name": "verificar_arquivos",
                    "description": "Nomeia todos os arquivos disponíveis na pasta atual e retorna como uma lista",
                    "parameters": {
                        "type": "object",
                        "properties": {},
                        "required": []
                    }         
                }
            }, 
            {
                "type": "function",
                "function": {
                    "name": "read_start_file",
                    "description": "Dado um arquivo selecionado ele lê o conteúdo inicial do arquivo",
                    "parameters": {
                        "type": "object",
                        "properties": {
                            "file_name": {
                                "type": "string",
                                "description": "Nome do arquivo a ser lido"
                            } 
                        },
                        "required": [ "file_name" ]
                    }
                }
            } 
        ]

        # Junta todas as listas de definições e remove duplicatas
        all_tools_combined = registred_tools + my_tools + tools_to_add_definitions
        unique_tools = {}
        for tool in all_tools_combined:
            nome_da_tool = tool["function"]["name"]
            unique_tools[nome_da_tool] = tool
            
        self.__dict__["tools_definitions"] = list(unique_tools.values())


        # === 2. REGISTRO DOS PONTEIROS DE EXECUÇÃO DAS FUNÇÕES ===
        # Registra as funções nativas (Garante que não vão ser apagadas)
        self.__dict__["tools"]["verificar_arquivos"] = verificar_arquivos
        self.__dict__["tools"]["read_start_file"] = read_start_file

        # Adiciona APENAS as novas ferramentas vindas de fora sem sobrescrever as nativas
        for nome_tool, func in tools_to_add.items():
            if nome_tool not in self.__dict__["tools"]:
                self.__dict__["tools"][nome_tool] = func

        return self.tools
        # return self.tools

    
    def format_system_prompt(self, prompt_base: str = None ) -> str:
        """Formata o system prompt específico para o modelo e o Retorna no final ."""
        
        self.system_prompt = """Você é um Agente Autônomo Especialista (nível Engenheiro Principal de Software), projetado para resolver problemas complexos de forma metódica, autônoma e implacável.

            Sua arquitetura mental é baseada no ciclo "Pensar -> Agir -> Avaliar". Para cada solicitação do usuário, você deve planejar estrategicamente antes de tomar decisões.

            ====================================================================
            DIRETRIZES DE EXECUÇÃO E RACIOCÍNIO (CHAIN-OF-THOUGHT)
            ====================================================================
            1. PLANEJAMENTO PRÉVIO: Antes de emitir a resposta final, avalie a complexidade do problema. Se o problema for complexo, divida-o em etapas lógicas menores.
            2. VERIFICAÇÃO DE FATOS (EVIDÊNCIAS): NUNCA presuma o conteúdo de um arquivo, a estrutura de um diretório ou o estado de um sistema. Se você não tem certeza absoluta, você DEVE usar as ferramentas disponíveis para investigar e confirmar os dados antes de prosseguir.
            3. ADAPTAÇÃO A ERROS: Se uma chamada de ferramenta retornar um erro ou um resultado inesperado, não entre em pânico e não desista. Analise a mensagem de erro logicamente, ajuste os parâmetros e faça uma nova tentativa.
            4. MINIMIZAÇÃO DE ALUCINAÇÕES: Se as ferramentas disponíveis não forem suficientes para resolver o problema, ou se você não tiver acesso à informação necessária, seja totalmente transparente com o usuário e explique exatamente o que está bloqueando o seu avanço.
            5. PROIBIÇÃO DE EXEMPLOS JSON: NUNCA escreva blocos de código JSON simulando chamadas de ferramentas no meio da sua resposta para o usuário. Se for explicar como uma ferramenta funciona, explique apenas em texto natural, sem usar a sintaxe {"name":...}.
            6. PROIBIÇÃO DE INVENÇÃO: Você só pode usar as ferramentas que foram explicitamente fornecidas a você. NUNCA invente nomes de ferramentas (como 'responder_usuario', 'final_answer', etc).
            7. COMO RESPONDER AO USUÁRIO: Quando você decidir que tem a resposta final, NÃO chame nenhuma ferramenta. Simplesmente escreva a sua resposta em texto puro (Markdown). A ausência de uma chamada de ferramenta é o gatilho que envia a sua mensagem ao usuário.
            
            ====================================================================
            DIRETRIZES DE USO DE FERRAMENTAS (TOOL CALLING)
            ====================================================================
            - Você possui acesso a um conjunto de ferramentas (Tools) perfeitamente integradas ao seu ambiente.
            - Utilize-as de forma proativa. Você pode (e deve) encadear múltiplas ferramentas em sequência se a tarefa exigir investigação profunda.
            - Escolha os argumentos das ferramentas com precisão cirúrgica, baseando-se estritamente no histórico da conversa e nos dados que você já coletou.

            ====================================================================
            DIRETRIZES DE COMUNICAÇÃO
            ====================================================================
            - Idioma: Responda sempre em Português do Brasil (pt-BR) de forma técnica, culta e direta.
            - Tom: Profissional, pragmático e focado em soluções. Evite saudações excessivas ou linguagem poética/emocional.
            - Formatação: Quando estiver redigindo a resposta final para o usuário, use formatação Markdown rica (tabelas, listas, blocos de código com linguagem definida) para tornar a informação o mais legível e estruturada possível.
            - Síntese: Ao entregar a solução final, referencie os dados que você descobriu usando suas ferramentas para embasar suas conclusões.

            Seu objetivo final é resolver a dor do usuário com o menor número de interações manuais possíveis, automatizando a investigação e a resolução através da sua inteligência e das suas ferramentas.
            """
        
        return self.system_prompt

    
    def execute_tool_chain(self, tools_calls : list ) -> dict:
        """Define como as ferramentas serão encadeadas e chamadas."""

        # Depois vou testar uma versão usando o id como key ao envez do nome
        results = { call["name"] : self.tools[call["name"]]( **call["arguments"] ) for call in tools_calls }

        return results 

    
    def parse_response(self, raw_response: str) -> List[Dict[str, Any]]:
        """
        Processa a saída bruta do modelo (JSON, texto, etc) e retorna uma lista 
        de todas as tool_calls, ignorando conversas e formatações Markdown.
        """
        raw_response = raw_response.content
        tools_encontradas = []
        if not raw_response or not raw_response.strip():
            return tools_encontradas

        # =====================================================================
        # ESTRATÉGIA 1: Buscar um Array de Objetos JSON (ex: [ {tool1}, {tool2} ])
        # =====================================================================
        match_lista = re.search(r'\[\s*\{.*?\}\s*\]', raw_response, re.DOTALL)
        if match_lista:
            try:
                lista_obj = json.loads(match_lista.group(0))
                if isinstance(lista_obj, list):
                    for obj in lista_obj:
                        if self._is_valid_tool_call(obj):
                            tools_encontradas.append(self._normalize_tool_call(obj))
                    
                    # Se encontrou tools válidas dentro da lista, encerra a busca
                    if tools_encontradas:
                        return tools_encontradas
            except json.JSONDecodeError:
                pass # Se falhar, segue para a estratégia 2

        # =====================================================================
        # ESTRATÉGIA 2: Varrer o texto pescando múltiplos JSONs soltos {...}
        # =====================================================================
        decoder = json.JSONDecoder()
        pos = 0
        
        while pos < len(raw_response):
            # Procura a posição de abertura de chave '{'
            match = re.search(r'\{', raw_response[pos:])
            if not match:
                break
                
            start_idx = pos + match.start()
            
            try:
                # O raw_decode extrai APENAS o JSON válido a partir do '{'
                obj, chars_consumed = decoder.raw_decode(raw_response[start_idx:])
                
                if self._is_valid_tool_call(obj):
                    tools_encontradas.append(self._normalize_tool_call(obj))
                    
                # Avança o cursor pulando todo o JSON que acabamos de extrair
                pos = start_idx + chars_consumed
                
            except json.JSONDecodeError:
                # Se o '{' não iniciava um JSON válido, avança 1 caractere
                pos = start_idx + 1

        return tools_encontradas

    # --- MÉTODOS AUXILIARES PRIVADOS (Boas Práticas de POO) ---

    def _is_valid_tool_call(self, obj: Any) -> bool:
        """Verifica se o dicionário possui a assinatura mínima de uma ferramenta."""
        return isinstance(obj, dict) and "name" in obj

    def _normalize_tool_call(self, obj: Dict[str, Any]) -> Dict[str, Any]:
        """
        Padroniza a estrutura da chamada, garantindo que 'arguments' exista
        e injetando um ID único simulando o comportamento nativo da API.
        """
        # Trata o caso onde o modelo manda 'arguments' como string (Padrão OpenAI) 
        # ou como dicionário (Vício de modelos locais)
        args = obj.get("arguments", {})
        if isinstance(args, str):
            try:
                args = json.loads(args)
            except json.JSONDecodeError:
                args = {} # Fallback de segurança se a string for inválida

        return {
            "id": f"call_{uuid.uuid4().hex[:8]}",
            "name": str(obj["name"]).strip(),
            "arguments": args
        }
    
    def build_response_of_tools_results( self, tool_responses : dict ) -> str :
        """Monta a resposta textual que consolida os resultados das chamadas de ferramentas"""
        if not tool_responses:
            return "<tool_results status='empty'>\n  <message>Nenhuma ação foi executada.</message>\n</tool_results>"

        xml_blocks = ["<tool_results status='success'>"]
        
        for nome_tool, resultado in tool_responses.items():
            
            # 1. Identificação do tipo real do dado (Metadado valioso para o LLM)
            tipo_dado = type(resultado).__name__
            
            # 2. Serialização segura
            if isinstance(resultado, (dict, list)):
                payload_str = json.dumps(resultado, indent=2, ensure_ascii=False)
            else:
                payload_str = str(resultado)
                
            # 3. Tratamento de payloads massivos (Truncamento defensivo opcional)
            # LLMs perdem o raciocínio se uma tool cuspir 50 mil linhas de log de uma vez.
            MAX_CHARS = 10000 
            if len(payload_str) > MAX_CHARS:
                payload_str = payload_str[:MAX_CHARS] + f"\n\n[... TRUNCADO: Excedeu {MAX_CHARS} caracteres. Refine a busca.]"
                status_exec = "partial_truncation"
            else:
                status_exec = "complete"

            # 4. Construção do bloco Semântico (Fronteira Rígida de Atenção)
            xml_blocks.append(f'  <tool_response name="{nome_tool}" execution_status="{status_exec}" data_type="{tipo_dado}">')
            xml_blocks.append(f'    <payload>')
            
            # O texto da payload recuado para manter a estrutura da árvore clara
            xml_blocks.append(f'      {payload_str.replace(chr(10), chr(10) + "      ")}')
            
            xml_blocks.append(f'    </payload>')
            xml_blocks.append(f'  </tool_response>')

        xml_blocks.append("</tool_results>")
        
        return "\n".join(xml_blocks)
    
class qwen3_6_Agent( AIAgentInterface ):
    def __init__(self , model = "qwen3.6:35b-a3b" ):
        super().__init__()
        self.model = model

    def tool_registry(self , tools_to_add : dict = {} , tools_to_add_definitions : list[dict] = [] ) -> list[dict]:
        """Dicionário obrigatório para mapear as ferramentas que podem ser chamadas pelo Agent."""
        
        # 0. Garante a existência segura das variáveis internas sem acionar o __getattr__
        if "tools_definitions" not in self.__dict__:
            self.__dict__["tools_definitions"] = []
        if "tools" not in self.__dict__:
            self.__dict__["tools"] = {}

        # === 1. REGISTRO DOS JSON SCHEMAS (DEFINIÇÕES) ===
        registred_tools = self.__dict__["tools_definitions"]
        
        my_tools = [verificar_arquivos.schema , read_start_file.schema ]
        # my_tools = [
        #     {
        #         "type": "function",
        #         "function": {
        #             "name": "verificar_arquivos",
        #             "description": "Nomeia todos os arquivos disponíveis na pasta atual e retorna como uma lista",
        #             "parameters": {
        #                 "type": "object",
        #                 "properties": {},
        #                 "required": []
        #             }         
        #         }
        #     }, 
        #     {
        #         "type": "function",
        #         "function": {
        #             "name": "read_start_file",
        #             "description": "Dado um arquivo selecionado ele lê o conteúdo inicial do arquivo",
        #             "parameters": {
        #                 "type": "object",
        #                 "properties": {
        #                     "file_name": {
        #                         "type": "string",
        #                         "description": "Nome do arquivo a ser lido"
        #                     } 
        #                 },
        #                 "required": [ "file_name" ]
        #             }
        #         }
        #     } 
        # ]

        # Junta todas as listas de definições e remove duplicatas
        all_tools_combined = registred_tools + my_tools + tools_to_add_definitions
        unique_tools = {}
        for tool_ in all_tools_combined:
            nome_da_tool = tool_["function"]["name"]
            unique_tools[nome_da_tool] = tool_
            
        self.__dict__["tools_definitions"] = list(unique_tools.values())


        # === 2. REGISTRO DOS PONTEIROS DE EXECUÇÃO DAS FUNÇÕES ===
        # Registra as funções nativas (Garante que não vão ser apagadas)
        self.__dict__["tools"]["verificar_arquivos"] = verificar_arquivos
        self.__dict__["tools"]["read_start_file"] = read_start_file

        # Adiciona APENAS as novas ferramentas vindas de fora sem sobrescrever as nativas
        for nome_tool, func in tools_to_add.items():
            if nome_tool not in self.__dict__["tools"]:
                self.__dict__["tools"][nome_tool] = func

        return self.tools
        # return self.tools

    
    def format_system_prompt(self, prompt_base: str = None ) -> str:
        """Formata o system prompt específico para o modelo e o Retorna no final ."""
        
        self.system_prompt = """Você é um Agente Autônomo Especialista (nível Engenheiro Principal de Software), projetado para resolver problemas complexos de forma metódica, autônoma e implacável.

            Sua arquitetura mental é baseada no ciclo "Pensar -> Agir -> Avaliar". Para cada solicitação do usuário, você deve planejar estrategicamente antes de tomar decisões.

            ====================================================================
            DIRETRIZES DE EXECUÇÃO E RACIOCÍNIO (CHAIN-OF-THOUGHT)
            ====================================================================
            1. PLANEJAMENTO PRÉVIO: Antes de emitir a resposta final, avalie a complexidade do problema. Se o problema for complexo, divida-o em etapas lógicas menores.
            2_A. VERIFICAÇÃO DE FATOS (EVIDÊNCIAS): NUNCA presuma o conteúdo de um arquivo, a estrutura de um diretório ou o estado de um sistema. Se você não tem certeza absoluta, você DEVE usar as ferramentas disponíveis para investigar e confirmar os dados antes de prosseguir.
            2_B. VERIFICAÇÃO DE FATOS (TOOL CALLS): A chamada de tool DEVE estar SEMPRE condicionada a estratégia que VOCÊ tenha justificado logicamente no seu reasoning , SE não conseguir justificar lógicamente a solicitação da tool então não a chame , SE você não conseguir ter certeza o suficiente ainda sobre qual estratégia seguir devolva uma pergunta pedindo mais informações objetivando decidir qual será a estratégia .
            3. ADAPTAÇÃO A ERROS: Se uma chamada de ferramenta retornar um erro ou um resultado inesperado, não entre em pânico e não desista. Analise a mensagem de erro logicamente, ajuste os parâmetros e faça uma nova tentativa.
            4. MINIMIZAÇÃO DE ALUCINAÇÕES: Se as ferramentas disponíveis não forem suficientes para resolver o problema, ou se você não tiver acesso à informação necessária, seja totalmente transparente com o usuário e explique exatamente o que está bloqueando o seu avanço.
            5. PROIBIÇÃO DE EXEMPLOS JSON: NUNCA escreva blocos de código JSON simulando chamadas de ferramentas no meio da sua resposta para o usuário. Se for explicar como uma ferramenta funciona, explique apenas em texto natural, sem usar a sintaxe {"name":...}.
            6. PROIBIÇÃO DE INVENÇÃO: Você só pode usar as ferramentas que foram explicitamente fornecidas a você. NUNCA invente nomes de ferramentas (como 'responder_usuario', 'final_answer', etc).
            7. COMO RESPONDER AO USUÁRIO: Quando você decidir que tem a resposta final, NÃO chame nenhuma ferramenta. Simplesmente escreva a sua resposta em texto puro (Markdown). A ausência de uma chamada de ferramenta é o gatilho que envia a sua mensagem ao usuário.
            
            ====================================================================
            DIRETRIZES DE USO DE FERRAMENTAS (TOOL CALLING)
            ====================================================================
            - Você possui acesso a um conjunto de ferramentas (Tools) perfeitamente integradas ao seu ambiente.
            - Utilize-as de forma proativa. Você pode (e deve) encadear múltiplas ferramentas em sequência se a tarefa exigir investigação profunda.
            - Escolha os argumentos das ferramentas com precisão cirúrgica, baseando-se estritamente no histórico da conversa e nos dados que você já coletou.

            ====================================================================
            DIRETRIZES DE COMUNICAÇÃO
            ====================================================================
            - Idioma: Responda sempre em Português do Brasil (pt-BR) de forma técnica, culta e direta.
            - Tom: Profissional, pragmático e focado em soluções. Evite saudações excessivas ou linguagem poética/emocional.
            - Formatação: Quando estiver redigindo a resposta final para o usuário, use formatação Markdown rica (tabelas, listas, blocos de código com linguagem definida) para tornar a informação o mais legível e estruturada possível.
            - Síntese: Ao entregar a solução final, referencie os dados que você descobriu usando suas ferramentas para embasar suas conclusões.

            Seu objetivo final é resolver a dor do usuário com o menor número de interações manuais possíveis, automatizando a investigação e a resolução através da sua inteligência e das suas ferramentas.
            """
        
        return self.system_prompt

    
    def execute_tool_chain(self, tools_calls : list ) -> dict:
        """Define como as ferramentas serão encadeadas e chamadas."""

        # Depois vou testar uma versão usando o id como key ao envez do nome
        results = { call["name"] : self.tools[call["name"]]( **call["arguments"] ) for call in tools_calls }

        return results 

    
    def parse_response(self, raw_response: str) -> List[Dict[str, Any]]:
        """
        Processa a saída bruta do modelo (JSON, texto, etc) e retorna uma lista 
        de todas as tool_calls, ignorando conversas e formatações Markdown.
        """
        # raw_response = raw_response.content
        # raw_tool_calls = raw_response.tool_calls

        if raw_response.tool_calls != None :
            tools_encontradas = [ {"id": call.id , "name": call.function.name , "arguments": json.loads(call.function.arguments) } for call in raw_response.tool_calls ]
        else :
            tools_encontradas = []
            
        # {
        #     "id": f"call_{uuid.uuid4().hex[:8]}",
        #     "name": str(obj["name"]).strip(),
        #     "arguments": args
        # }
        


        return tools_encontradas

    # --- MÉTODOS AUXILIARES PRIVADOS (Boas Práticas de POO) ---

    def _is_valid_tool_call(self, obj: Any) -> bool:
        """Verifica se o dicionário possui a assinatura mínima de uma ferramenta."""
        return isinstance(obj, dict) and "name" in obj

    def _normalize_tool_call(self, obj: Dict[str, Any]) -> Dict[str, Any]:
        """
        Padroniza a estrutura da chamada, garantindo que 'arguments' exista
        e injetando um ID único simulando o comportamento nativo da API.
        """
        # Trata o caso onde o modelo manda 'arguments' como string (Padrão OpenAI) 
        # ou como dicionário (Vício de modelos locais)
        args = obj.get("arguments", {})
        if isinstance(args, str):
            try:
                args = json.loads(args)
            except json.JSONDecodeError:
                args = {} # Fallback de segurança se a string for inválida

        return {
            "id": f"call_{uuid.uuid4().hex[:8]}",
            "name": str(obj["name"]).strip(),
            "arguments": args
        }
    
    def build_response_of_tools_results( self, tool_responses : dict ) -> str :
        """Monta a resposta textual que consolida os resultados das chamadas de ferramentas"""
        if not tool_responses:
            return "<tool_results status='empty'>\n  <message>Nenhuma ação foi executada.</message>\n</tool_results>"

        xml_blocks = ["<tool_results status='success'>"]
        
        for nome_tool, resultado in tool_responses.items():
            
            # 1. Identificação do tipo real do dado (Metadado valioso para o LLM)
            tipo_dado = type(resultado).__name__
            
            # 2. Serialização segura
            if isinstance(resultado, (dict, list)):
                payload_str = json.dumps(resultado, indent=2, ensure_ascii=False)
            else:
                payload_str = str(resultado)
                
            # 3. Tratamento de payloads massivos (Truncamento defensivo opcional)
            # LLMs perdem o raciocínio se uma tool cuspir 50 mil linhas de log de uma vez.
            MAX_CHARS = 10000 
            if len(payload_str) > MAX_CHARS:
                payload_str = payload_str[:MAX_CHARS] + f"\n\n[... TRUNCADO: Excedeu {MAX_CHARS} caracteres. Refine a busca.]"
                status_exec = "partial_truncation"
            else:
                status_exec = "complete"

            # 4. Construção do bloco Semântico (Fronteira Rígida de Atenção)
            xml_blocks.append(f'  <tool_response name="{nome_tool}" execution_status="{status_exec}" data_type="{tipo_dado}">')
            xml_blocks.append(f'    <payload>')
            
            # O texto da payload recuado para manter a estrutura da árvore clara
            xml_blocks.append(f'      {payload_str.replace(chr(10), chr(10) + "      ")}')
            
            xml_blocks.append(f'    </payload>')
            xml_blocks.append(f'  </tool_response>')

        xml_blocks.append("</tool_results>")
        
        return "\n".join(xml_blocks)
    

# class deepseekAgent( AIAgentInterface ):
```
